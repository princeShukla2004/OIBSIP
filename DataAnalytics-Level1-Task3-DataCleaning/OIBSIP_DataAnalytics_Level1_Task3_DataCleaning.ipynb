## Missing Data Handling Strategy

The dataset contains missing values in the Age, Cabin, and Embarked columns.

### Age
Missing Age values will be replaced using the median age.

**Reason:** Age is a numerical variable and the median is less affected by extreme values than the mean.

### Embarked
Missing Embarked values will be replaced using the mode.

**Reason:** Embarked is a categorical variable, so the most frequent category is an appropriate choice.

### Cabin
The Cabin column contains a large number of missing values. Instead of deleting rows, missing Cabin values will be represented as "Unknown".

**Reason:** Deleting rows with missing Cabin values would result in a significant loss of data. The missing Cabin information can also represent the fact that the cabin information was not available.

```python
# Check missing values before cleaning

print("Missing values before handling:")
print(df[["Age", "Cabin", "Embarked"]].isnull().sum())
```

```python
# Handle missing values

# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with the mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Replace missing Cabin values with "Unknown"
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing values after handling:")
print(df[["Age", "Cabin", "Embarked"]].isnull().sum())
```

```python
# Count duplicate rows before removal

duplicates_before = df.duplicated().sum()

print("Duplicate Rows Before Cleaning:", duplicates_before)
```

```python
# Remove duplicate rows

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicate Rows After Cleaning:", duplicates_after)
```

## Duplicate Removal

Duplicate rows were checked before the cleaning process.

```python
# Check unique values in categorical columns

print("Sex values:")
print(df["Sex"].unique())

print("\nEmbarked values:")
print(df["Embarked"].unique())

print("\nPclass values:")
print(df["Pclass"].unique())
```

```python
# Standardise categorical text values

df["Sex"] = df["Sex"].astype(str).str.strip().str.lower()
df["Sex"] = df["Sex"].replace({
    "male": "Male",
    "female": "Female"
})

df["Embarked"] = df["Embarked"].astype(str).str.strip().str.upper()

print("Standardised Sex values:")
print(df["Sex"].unique())

print("\nStandardised Embarked values:")
print(df["Embarked"].unique())
```

## Data Standardisation

Categorical text values were checked for inconsistent formatting.

The `Sex` column was standardised to consistent values such as `Male` and `Female`.

The `Embarked` column was standardised using uppercase category codes.

The dataset does not contain a date column, so date-format standardisation was not applicable.

```python
# Detect outliers using the IQR method

numeric_columns = ["Age", "Fare", "SibSp", "Parch"]

outlier_report = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_report.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

display(outlier_report.round(2))
```

## Outlier Detection and Handling

The Interquartile Range (IQR) method was used to identify potential outliers in the numeric columns `Age`, `Fare`, `SibSp`, and `Parch`.

A value was considered a potential outlier when it was below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR.

The identified outliers were reviewed as potential extreme values rather than automatically treating them as data errors.

Extreme values in variables such as Fare, SibSp, and Parch can represent genuine passenger characteristics. Therefore, legitimate extreme observations were retained instead of being removed.

No artificial values were created or removed solely to reduce the number of outliers.

```python
# Correct data types

df["PassengerId"] = df["PassengerId"].astype(str)
df["Survived"] = df["Survived"].astype(int)
df["Pclass"] = df["Pclass"].astype(int)
df["Age"] = df["Age"].astype(float)
df["SibSp"] = df["SibSp"].astype(int)
df["Parch"] = df["Parch"].astype(int)
df["Fare"] = df["Fare"].astype(float)

df["Sex"] = df["Sex"].astype(str)
df["Embarked"] = df["Embarked"].astype(str)
df["Name"] = df["Name"].astype(str)
df["Ticket"] = df["Ticket"].astype(str)
df["Cabin"] = df["Cabin"].astype(str)

print("Data types after correction:")
print(df.dtypes)
```
